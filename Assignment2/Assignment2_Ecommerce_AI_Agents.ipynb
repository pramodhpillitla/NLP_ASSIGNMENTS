### 3. Multi-Turn Interactive Agent Demonstration
We simulate a complete end-to-end customer journey:
- Search and product discovery
- Cart management & inventory validation
- Applying promo discount coupons
- Policy inquiry (returns & refunds)
- Transactional checkout
- Post-purchase order tracking



```python
agent = EcommerceMultiAgentSystem()
session = "notebook_session_b210195"

queries = [
    "I am looking for a lightweight laptop for college under $1000 with great battery life.",
    "Add the laptop to my shopping cart.",
    "Can you apply coupon code FESTIVE20?",
    "What is your store return and refund policy on laptops?",
    "Proceed to checkout please.",
    "Can you track my previous order ORD-9821?"
]

for q in queries:
    print(f"\n>>> USER: {q}")
    trace = agent.route_and_execute(q, session_id=session)
    print(f"[Agent Intent]: {trace['intent']}")
    print(f"[ReAct Thought]: {trace['thought']}")
    print(f"[Tool Executed]: {trace['action']}")
    print(f"[Response]:\n{trace['final_response']}\n" + "-"*50)

```
